```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(20,1, -1)
y = np.arange(1,20)
```

```python

l, = plt.plot(x,y)
plt.plot(x,y)
t = plt.title("1")
plt.plot(x*2, y/2)
l.set_color("m")
l.set_data(x+2, y+2)
t.set_size(109)
```

```python
fig, (ax1, ax2) =  plt.subplots(ncols=2)
ax1.plot(x,y, label="one")
ax2.plot(y*3, x*3, label="two")
ax1.legend()
ax2.plot(x-1,y-1, label="three")
ax1.plot(y+2, x+2, label="four")
ax2.legend()
ax1.legend()
fig.legend()
```
